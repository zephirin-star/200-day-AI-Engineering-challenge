```python
records = [12.5, 8.0, 12.5, 15.2, 8.0, 20.0]

def find_duplicates(records):
  seen = []
  duplicates = []
  for x in records :
    if x not in seen :
      seen.append(x)
    elif x not in duplicates:
        duplicates.append(x)
  return duplicates
  print(find_duplicates(records))
```

```python
#Test 1
records = [1, 2, 1, 3, 2]
assert find_duplicates([1, 2, 1, 3, 2]) == [1, 2]
assert find_duplicates([1, 2, 3]) == []
assert find_duplicates([]) == []
assert find_duplicates([5, 5, 5]) ==[5]
print ("Test 1 passed!")
```

```python
#Test 2 the original list remains unchanged
records = [2, 5, 3, 3, 7]
original_records = records.copy()
find_duplicates(records)
assert records == original_records
print("original list unchanged!")
```

## Day_10_Duplicate_Detection
- The function find_duplication(records) creates two lists seen and duplicates
- seen stores each unique value encountered while looping through the records.
- duplicates stores values that appear again, but each duplicate is added only once. into duplicates
