```python
def create_quality_report(audit_result):
    """ Analyze an audit result and return
     a data quality summary dictionary."""
    valid_distances = audit_result.get("valid_distances", [])
    rejected_rows = audit_result.get("rejected_rows", 0)

    # Calculate total_val_dist manually (without len)
    total_val_dist = 0
    for distance in valid_distances:
        total_val_dist += 1

    total_rows = rejected_rows + total_val_dist

    if total_rows == 0:
        quality_percentage = 0.0
    else:
        quality_percentage = (total_val_dist / total_rows) * 100

    # Calculate total distance manually (without sum)
    total_distance = 0
    for distance in valid_distances:
        total_distance += distance

    if total_val_dist > 0:
        average_valid_distance = total_distance / total_val_dist
    else:
        average_valid_distance = None

    report = {
        "total_rows": total_rows,
        "valid_rows": total_val_dist,
        "rejected_rows": rejected_rows,
        "quality_percentage": quality_percentage,
        "average_valid_distance": average_valid_distance
    }
    return report

# Test the function with your the dictionary
audit_result_dict = {
    "valid_distances": [5, 10, 0, 8, 7],
    "rejected_rows": 3,
    "total_miles": 30
}

report = create_quality_report(audit_result_dict)
print(report)
```

```python
# Test 1 - Normal dataset
audit = {
    "valid_distances": [5, 10, 0, 8, 7],
    "rejected_rows": 3,
    "total_miles": 30
}

report = create_quality_report(audit)

assert report["total_rows"] == 8
assert report["valid_rows"] == 5
assert report["rejected_rows"] == 3
assert report["quality_percentage"] == 62.5
assert report["average_valid_distance"] == 6.0

print("Test 1 passed!")
```

```python
#Test 2-Empty dataset
empty = {
    "valid_distances": [],
    "rejected_rows": 0,
    "total_miles": 0
}

report = create_quality_report(empty)

assert report["total_rows"] == 0
assert report["quality_percentage"] == 0.0
assert report["average_valid_distance"] is None

print("Test 2 passed!")
```

```python
# Test 3 - Original input unchanged
audit = {
    "valid_distances": [2, 4, 6],
    "rejected_rows": 1,
    "total_miles": 12
}

original = {
    "valid_distances": audit["valid_distances"].copy(),
    "rejected_rows": audit["rejected_rows"],
    "total_miles": audit ["total_miles"]
}

create_quality_report(audit)

assert audit == original

print("Test 3 passed!")
```

### Day 8 — Data Quality Report

The function `create_quality_report()` analyzes a dataset audit and produces a data quality summary.

It:
- Counts valid and rejected rows.
- Calculates the total number of records.
- Calculates the percentage of valid records.
- Calculates the average valid distance.
- Handles empty datasets.
- Returns the results as a dictionary without modifying the original input.

**Tests performed:**
- Normal dataset
- Empty dataset
- Original input remains unchanged
