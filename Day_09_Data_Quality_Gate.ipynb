```python
def evaluate_data_quality(report, threshold=80.0):
  """
  Decide whether a dataset passes a quality threshold.
  """
  total_rows = report.get("total_rows", 0)
  quality_percentage = report.get("quality_percentage", 0.0)

  if total_rows == 0:
    return "EMPTY"

  if quality_percentage >= threshold:
    return "PASS"
  else:
    return "REJECT"
```

```python
# Test1
report = {
    "total_rows": 8,
    "quality_percentage": 62.5
}
assert evaluate_data_quality(report) == "REJECT"

# Test 2 Exactly at threshold
report = {
    "total_rows": 10,
    "quality_percentage": 80.0
}
assert evaluate_data_quality(report)== "PASS"
# Test 3
report = {
    "total_rows": 0,
    "quality_percentage": 0.0
}
assert evaluate_data_quality(report)== "EMPTY"
# Test 4
report = {
    "total_rows": 10,
    "quality_percentage": 75.0
 }
assert evaluate_data_quality(report, 70.0)== "PASS"
 # Test 5- Input unchanged
report = {
     "total_rows": 10,
     "quality_percentage": 90.0
 }

original = report.copy()
evaluate_data_quality(report)

assert report == original

print("All Day 9 tests passed!")
```
