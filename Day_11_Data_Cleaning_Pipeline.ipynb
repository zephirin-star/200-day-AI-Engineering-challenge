```python
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]

def clean_delivery_pipeline(records):
    duplicates = []
    cleaned_records = []
    rejected_count = 0

    for record in records:
        # Clean up string formatting (e.g., replace commas with periods)
        cleaned_str = str(record).replace(',', '.')

        try:
            val = float(cleaned_str)
        except ValueError:
            # If it cannot be converted to a float (like 'unknown'), count as rejected
            rejected_count += 1
            continue

        if val <= 0:
            rejected_count += 1
        elif val in cleaned_records:
           if val not in duplicates:
            duplicates.append(val)
        else:
            cleaned_records.append(val)

    report = {
        "cleaned_records": cleaned_records,
        "rejected_count": rejected_count,
        "duplicates": duplicates
    }
    return report

# Tests
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]
result = clean_delivery_pipeline(records)
print(result)
```

```python
# Test
assert clean_delivery_pipeline(["5", "5", "5", "5"])["cleaned_records"] == [5.0]

assert clean_delivery_pipeline(["5", "5", "5", "5"])["duplicates"] == [5.0]

records = ["12.5", "-4", "12.5", "unknown"]
original = records.copy()

clean_delivery_pipeline(records)
assert records == original

print("All final tests passed!")
```
