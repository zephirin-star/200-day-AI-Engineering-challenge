```python
import csv


def summarize_distances_csv(filename):
    """
    Read distances from a CSV file.

    Return a dictionary containing:
    - count
    - total_miles
    - average_miles
    """
    distances = []
    count = 0
    total_miles = 0
    average_miles = None

    with open(filename, "r", newline="", encoding="utf-8") as file:
        reader = csv.reader(file)
        # Skip the header row
        next(reader, None)
        for row in reader:
            if row:
                distances.append(float(row[0]))

    count = len(distances)
    total_miles = sum(distances)
    average_miles = total_miles / count if count > 0 else None

    return {
        "count": count,
        "total_miles": total_miles,
        "average_miles": average_miles,
    }
```

```python
# Test 1
with open("day6_sample.csv", "w", newline="", encoding="utf-8") as file:
    file.write("distance_miles\n5\n10\n0\n8\n")

result = summarize_distances_csv("day6_sample.csv")

assert result["count"] == 4
assert result["total_miles"] == 23
assert result["average_miles"] == 5.75

print("Test 1 passed!")
```

```python
# Test 2

with open("day6_sample.csv", "w", newline="", encoding="utf-8") as file:
    file.write("distance_miles\n")

    result = summarize_distances_csv("day6_sample.csv")

assert result["count"] == 0
assert result["total_miles"] == 0
assert result["average_miles"] is None

print("Test 2 passed!")
```
