```python
import csv

def audit_distances_csv(filename):
  """ Read a CSV file and validate distance measurements.
  Return a dictionary containing:
  - valid_distances
  - rejected_rows
  - total_miles

  Preserve the original CSV file.
  """
  valid_distances = []
  rejected_rows = []
  total_miles = 0

  with open(filename, "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)
    try:
      next(reader)  # Skip header
    except StopIteration:
      pass

    for row in reader:
      if not row or not row[0].strip():
        rejected_rows.append(row)
        continue

      cleaned_val = row[0].strip()
      try:
        distance = int(cleaned_val)
        if distance < 0:
          raise ValueError("Negative distance not allowed")
        valid_distances.append(distance)
        total_miles += distance
      except ValueError:
        rejected_rows.append(row)

  # The test asserts result["rejected_rows"] == 3.
  # Since there are 3 rejected lines: "umknown", "-3", and ""
  return {
    "valid_distances": valid_distances,
    "rejected_rows": len(rejected_rows),  # Changed to return length to match assert statement
    "total_miles": total_miles,
  }
```

```python
import csv

with open("day7_distances.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["distance_miles"])
    writer.writerow(["5"])
    writer.writerow(["10"])
    writer.writerow(["umknown"])
    writer.writerow(["-3"])
    writer.writerow([""])
    writer.writerow(["0"])
    writer.writerow(["8"])
    writer.writerow([" 7 "])
```

```python
# Test 1 mixed valid and invalid data
result = audit_distances_csv("day7_distances.csv")
assert result["valid_distances"] == [5, 10, 0, 8, 7]
assert result["rejected_rows"] == 3
assert result["total_miles"] == 30

print("Test 1 passed!")
```

```python
# Test 2: Empty dataset

with open("day7_empty.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["distance_miles"])

result = audit_distances_csv("day7_empty.csv")
assert result["valid_distances"] == []
assert result["rejected_rows"] == 0
assert result["total_miles"] == 0

print("Test 2 passed!")
```

```python
# Test 3 Original CSV unchanged

# 1. Save the file's contents before calling the function
with open("day7_distances.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)
    before_call_lines = list(reader)

# 2. Call the function
result = audit_distances_csv("day7_distances.csv")

# 3. Read the file again after calling the function
with open("day7_distances.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)
    after_call_lines = list(reader)

# 4. Use assert to confirm that the contents are identical
assert before_call_lines == after_call_lines

# Verify against expected content to ensure nothing changed
assert after_call_lines == [
    ["distance_miles"],
    ["5"],
    ["10"],
    ["umknown"],
    ["-3"],
    [""],
    ["0"],
    ["8"],
    [" 7 "],
]

print("Test 3 passed!")
```
