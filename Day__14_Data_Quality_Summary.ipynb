```python
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]

def clean_delivery_pipeline(records):
    duplicates = []
    cleaned_records = []
    rejected_count = 0

    for record in records:
        # Clean up string formatting (e.g., replace commas with periods)
        cleaned_str = str(record).replace(',', '.')

        try:
            val = float(cleaned_str)
        except ValueError:
            rejected_count += 1
            continue

        if val <= 0:
            rejected_count += 1
        elif val in cleaned_records:
           if val not in duplicates:
            duplicates.append(val)
        else:
            cleaned_records.append(val)

    report = {
        "cleaned_records": cleaned_records,
        "rejected_count": rejected_count,
        "duplicates": duplicates
    }
    return report

# Tests
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]
result = clean_delivery_pipeline(records)
print(result)
```

```python
def summarize_data_quality(records, report):
    total_records = len(records)
    cleaned_count = len(report["cleaned_records"])
    rejected_count = report["rejected_count"]
    duplicate_count = len(report["duplicates"])

    try:
        quality_percentage = (cleaned_count / total_records) * 100
    except ZeroDivisionError:
        quality_percentage = 0.0

    summary_report = {
        "total_records": total_records,
        "cleaned_count": cleaned_count,
        "rejected_count": rejected_count,
        "duplicate_count": duplicate_count,
        "quality_percentage": quality_percentage
    }
    return summary_report

# 'result' contains the dictionary returned from clean_delivery_pipeline
final_summary = summarize_data_quality(records, result)
print(final_summary)
```

```python
# Test   the empty  dataset
empty_records = []
empty_report = clean_delivery_pipeline(empty_records)

empty_summary = summarize_data_quality(empty_records, empty_report)
assert empty_summary["total_records"] == 0.0
print("Test1 passed!")


# Test check input preservation

import copy

original_records = records.copy()
original_report = copy.deepcopy(result)

summarize_data_quality(records, result)

assert records == original_records
assert result == original_report
print("Test2 passed!")
```

## Day 14 — Data Quality Summary

### Purpose
The `summarize_data_quality()` function generates a summary of the quality of a delivery dataset.

### Inputs
The function accepts two parameters:
- `records`: The original list of delivery records.
- `report`: A dictionary returned by `clean_delivery_pipeline()`.

### Process
The function:
1. Counts the total number of records.
2. Counts the unique cleaned records.
3. Retrieves the number of rejected records.
4. Counts the unique duplicate values.
5. Calculates the quality percentage.
6. Handles an empty dataset to prevent division by zero.

### Output
The function returns a new dictionary containing `total_records`, `cleaned_count`, `rejected_count`, `duplicate_count`, and `quality_percentage`.

### Testing
The function was tested with a delivery dataset containing valid, invalid, and duplicate records. Additional tests will verify empty inputs and ensure that the original data remains unchanged.
