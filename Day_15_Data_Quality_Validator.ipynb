```python
import math

def validate_summary_report(summary):
  if not isinstance(summary, dict):
    return False

  required_keys = [
    "total_records",
    "cleaned_count",
    "rejected_count",
    "duplicate_count",
    "quality_percentage"
  ]

  # Check that all required keys are present
  for key in required_keys:
    if key not in summary:
      return False

  # 1. Type and value validation for integer count fields
  for key in ["total_records", "cleaned_count", "rejected_count", "duplicate_count"]:
    val = summary[key]
    # In Python, isinstance(True, int) is True, so we must explicitly check for bool
    if isinstance(val, bool) or not isinstance(val, int):
      return False
    if val < 0:
      return False

  # 2. Type and value validation for quality percentage
  qp = summary["quality_percentage"]
  if isinstance(qp, bool) or not isinstance(qp, (int, float)):
    return False
  if math.isnan(qp):
    return False
  if not (0 <= qp <= 100):
    return False

  return True

valid_summary = {
    "total_records": 7,
    "cleaned_count": 3,
    "rejected_count": 2,
    "duplicate_count": 2,
    "quality_percentage": 42.86
}

original = valid_summary.copy()
validate_summary_report(original)
assert valid_summary == original
print("passed!")
```
