```python
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]

def clean_delivery_pipeline(records):
    duplicates = []
    cleaned_records = []
    rejected_count = 0

    for record in records:
        # Clean up string formatting (e.g., replace commas with periods)
        cleaned_str = str(record).replace(',', '.')

        try:
            val = float(cleaned_str)
        except ValueError:
            rejected_count += 1
            continue

        if val <= 0:
            rejected_count += 1
        elif val in cleaned_records:
           if val not in duplicates:
            duplicates.append(val)
        else:
            cleaned_records.append(val)

    report = {
        "cleaned_records": cleaned_records,
        "rejected_count": rejected_count,
        "duplicates": duplicates
    }
    return report

# Tests
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]
result = clean_delivery_pipeline(records)
print(result)
```

```python
# Repeated record
records = [ "5", "5", "5", "5"]
result = clean_delivery_pipeline(records)
assert result["cleaned_records"] == [5.0]
assert result["rejected_count"] == 0
assert result["duplicates"] == [5.0]

print("All tests are passed!")
```

```python
# Test duplicates detection
records = ["10", "unknown", "-5", "0", "10"]
result = clean_delivery_pipeline(records)

assert result["cleaned_records"] == [10.0]

assert result["rejected_count"] == 3

assert result["duplicates"] == [10.0]

print("All tests are passed!")
```

```python
# Test empty datasets
records = []

result = clean_delivery_pipeline(records)

assert result["cleaned_records"] == []

assert result["rejected_count"] == 0

assert result["duplicates"] == []

print("All tests are passed!")
```

```python
# Test input preservation

records = ["12.5", "unknown", "12.5"]

original = records.copy()

result = clean_delivery_pipeline(records)

assert result["cleaned_records"] == [12.5]

assert result["rejected_count"] == 1

assert result["duplicates"] == [12.5]

assert records == original

print("All tests are passed!")
```
