```python
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]

def clean_delivery_pipeline(records):
    duplicates = []
    cleaned_records = []
    rejected_count = 0

    for record in records:
        # Clean up string formatting (e.g., replace commas with periods)
        cleaned_str = str(record).replace(',', '.')

        try:
            val = float(cleaned_str)
        except ValueError:
            rejected_count += 1
            continue

        if val <= 0:
            rejected_count += 1
        elif val in cleaned_records:
           if val not in duplicates:
            duplicates.append(val)
        else:
            cleaned_records.append(val)

    report = {
        "cleaned_records": cleaned_records,
        "rejected_count": rejected_count,
        "duplicates": duplicates
    }
    return report

# Tests
records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]
result = clean_delivery_pipeline(records)
print(result)
```

```python
import matplotlib.pyplot as plt

records = ["12.5", "8,0", "12.5", "-4", "unknown", "15.2", "8.0"]

report = clean_delivery_pipeline(records)


def visualize_data_quality(report):
  cleaned_records = report.get('cleaned_records', [])
  duplicates = report.get('duplicates', [])
  rejected_count = report.get('rejected_count', 0)
  cleaned = len(cleaned_records)
  duplicates = len(duplicates)

  categories = ["cleaned", "duplicates", "rejected_count"]
  counts = [cleaned, duplicates, rejected_count]

  plt.bar(categories, counts)
  plt.xlabel("Data Quality")
  plt.ylabel("Count")
  plt.title("Data Quality Visualization")
  plt.show()
  print("Data Quality Visualization")

visualize_data_quality(report)
```

## Day 13 — Data Quality Visualization

The `visualize_data_quality(report)` function extracts information from a report dictionary and displays it using a Matplotlib bar chart.

### What the function does

- Displays three bars representing cleaned records, duplicate values, and rejected records.
- Shows the category names on the X-axis and their corresponding counts on the Y-axis.
- Uses the report dictionary to determine the height of each bar.
